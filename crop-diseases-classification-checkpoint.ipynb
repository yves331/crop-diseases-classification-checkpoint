```python
def extract_features(image_path):
    """
    Transforme une image en vecteur numérique (liste de features).
    
    Paramètre:
        image_path (str) : chemin vers le fichier image
        
    Retourne:
        numpy.array : vecteur de features, ou None si l'image est illisible
    """
    # Lecture de l'image avec OpenCV (format BGR par défaut)
    img = cv2.imread(image_path)
    
    # Vérification : si l'image n'a pas pu être lue, on retourne None
    if img is None:
        return None
    
    features = []   # On va accumuler toutes les features dans cette liste

    # ── Feature 1 : Histogramme de couleurs (3 canaux × 32 bins) ──────────────
    # Un histogramme compte combien de pixels ont chaque niveau d'intensité.
    # On utilise 32 "bacs" (bins) entre 0 et 255 pour chaque canal de couleur.
    # Exemple : beaucoup de pixels avec une forte intensité rouge → plante stressée ?
    for i in range(3):   # i=0 : Bleu, i=1 : Vert, i=2 : Rouge (ordre BGR d'OpenCV)
        hist = cv2.calcHist([img], [i], None, [32], [0, 256])
        features.extend(hist.flatten())   # On aplatit le tableau 2D en liste 1D
    
    # ── Feature 2 : Moments statistiques par canal (moyenne, écart-type, médiane) ─
    # Ces statistiques résument la distribution des couleurs de façon très compacte.
    # Exemple : une plante chlorotique (jaune) aura moins de vert → moyenne canal vert faible
    for i in range(3):
        channel = img[:, :, i]   # Extraction du canal i (matrice 2D de pixels)
        features.extend([
            np.mean(channel),    # Valeur moyenne des pixels du canal
            np.std(channel),     # Écart-type (mesure de la variabilité)
            np.median(channel)   # Valeur médiane (robuste aux valeurs extrêmes)
        ])
    
    # ── Feature 3 : Texture en niveaux de gris ──────────────────────────────────
    # On convertit en niveaux de gris pour analyser la texture indépendamment de la couleur.
    # Une texture irrégulière (taches, lésions) aura un écart-type et une variance élevés.
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    features.extend([
        np.mean(gray),   # Luminosité moyenne
        np.std(gray),    # Variabilité de la luminosité (texture rugueuse = std élevé)
        np.var(gray)     # Variance (carré de l'écart-type)
    ])
    
    # ── Feature 4 : Détection de contours (algorithme de Canny) ─────────────────
    # Canny détecte les bords et transitions de l'image (contours de taches, nervures…).
    # On mesure la "densité" des contours = proportion de pixels qui sont des contours.
    # Une plante très malade peut avoir beaucoup de contours irréguliers.
    edges = cv2.Canny(gray, 50, 150)   # 50 et 150 : seuils bas et haut de détection
    features.append(np.sum(edges) / edges.size)   # Proportion de pixels "contour"
    
    return np.array(features)   # On retourne un tableau numpy de dimension fixe
```

```python
def load_label_map(data_path=DATA_PATH):
    """
    Charge le dictionnaire qui associe chaque numéro de classe à un nom de maladie.
    
    Exemple de contenu du JSON :
        {"0": "Bacterial Blight", "1": "Brown Spot", "2": "Healthy", ...}
    
    Retourne:
        dict : {int/str → str} mapping numéro de classe → nom de maladie
    """
    json_path = os.path.join(data_path, 'label_nom_to_disease_map.json')
    with open(json_path, 'r') as f:
        label_map = json.load(f)
    
    print(f"✅ {len(label_map)} classes détectées :")
    for k, v in label_map.items():
        print(f"   Classe {k} → {v}")
    
    return label_map


def load_data(data_path=DATA_PATH, sample_size=None):
    """
    Charge les images, extrait leurs features et construit X et y.
    
    Paramètres:
        data_path   : chemin vers le dossier Data/
        sample_size : si spécifié, ne traite que N images (utile pour tester rapidement)
    
    Retourne:
        X (numpy.array) : matrice de features, forme (N, nb_features)
        y (numpy.array) : vecteur de labels,   forme (N,)
    """
    print("📂 Chargement des données...")
    
    # Lecture du fichier CSV qui liste les images et leurs labels
    # Colonnes attendues : 'image_id' (nom du fichier) et 'label' (numéro de classe)
    train_df = pd.read_csv(os.path.join(data_path, 'train.csv'))
    print(f"   {len(train_df)} lignes dans train.csv")
    
    # Optionnel : prendre un sous-échantillon aléatoire pour aller plus vite
    if sample_size:
        train_df = train_df.sample(n=min(sample_size, len(train_df)), random_state=42)
        print(f"   → Sous-échantillon de {len(train_df)} images sélectionné")
    
    X = []   # Contiendra les vecteurs de features
    y = []   # Contiendra les labels correspondants
    
    print(f"⚙️  Extraction des features de {len(train_df)} images...")
    
    # Boucle sur chaque ligne du CSV (= chaque image)
    for idx, row in tqdm(train_df.iterrows(), total=len(train_df)):
        img_path = os.path.join(data_path, 'train_images', row['image_id'])
        
        # Sécurité : vérifier que l'image existe bien sur le disque
        if not os.path.exists(img_path):
            continue   # On passe à l'image suivante
        
        # Extraction des features de cette image
        features = extract_features(img_path)
        
        # On n'ajoute que si l'extraction a réussi (features != None)
        if features is not None:
            X.append(features)
            y.append(row['label'])
    
    print(f"\n✅ Dataset prêt : {len(X)} images × {len(X[0])} features")
    return np.array(X), np.array(y)
```

```python
# ══════════════════════════════════════════════════════════════════════
# VISUALISATION DE LA DISTRIBUTION DES CLASSES
# ══════════════════════════════════════════════════════════════════════

def plot_class_distribution(y_train, y_test, label_map, figsize=(14, 6)):
    """
    Affiche la distribution du nombre d'images par classe dans train et test.
    
    Permet de détecter un éventuel déséquilibre entre les classes
    (ex : 2000 images d'une classe et seulement 50 d'une autre).
    """
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    labels = list(label_map.values())
    # labels = list(label_map.values())
    colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A', '#98D8C8']
    
    for ax, y_data, title_prefix in zip(axes, [y_train, y_test], ["Train", "Test"]):
        counts = pd.Series(y_data).value_counts().sort_index()
        ax.bar(range(len(counts)), counts.values,
               color=colors[:len(counts)], alpha=0.8, edgecolor='black', linewidth=1.5)
        ax.set_title(f'Distribution {title_prefix} (Total: {len(y_data)})',
                     fontsize=14, fontweight='bold')
        ax.set_xlabel('Classes', fontsize=12)
        ax.set_ylabel("Nombre d'images", fontsize=12)
        ax.set_xticks(range(len(labels)))
        ax.set_xticklabels(labels, rotation=45, ha='right', fontsize=10)
        ax.grid(axis='y', alpha=0.3, linestyle='--')
        # Afficher les valeurs au-dessus de chaque barre
        for i, v in enumerate(counts.values):
            ax.text(i, v + max(counts.values) * 0.01, str(v),
                    ha='center', va='bottom', fontweight='bold', fontsize=10)
    
    plt.tight_layout()
    plt.show()
    
    # Tableau récapitulatif
    print("\n" + "="*70)
    print("📊 DISTRIBUTION DÉTAILLÉE DES CLASSES")
    print("="*70)
    print(f"{'Classe':<40} {'Train':<15} {'Test':<15} {'Total':<10}")
    print("-"*70)
    
    train_counts = pd.Series(y_train).value_counts().sort_index()
    test_counts  = pd.Series(y_test).value_counts().sort_index()
    
    for i, label in enumerate(labels):
        n_train = train_counts.get(i, 0)
        n_test  = test_counts.get(i, 0)
        pct_train = n_train / len(y_train) * 100 if len(y_train) > 0 else 0
        pct_test  = n_test  / len(y_test)  * 100 if len(y_test)  > 0 else 0
        print(f"{label:<40} {n_train:<6} ({pct_train:>5.1f}%)  "
              f"{n_test:<6} ({pct_test:>5.1f}%)  {n_train+n_test:<10}")
    
    print("-"*70)
    print(f"{'TOTAL':<40} {len(y_train):<15} {len(y_test):<15} "
          f"{len(y_train)+len(y_test):<10}")
    print("="*70)
```

```python
# ══════════════════════════════════════════════════════════════════════
# ÉVALUATION D'UN MODÈLE
# ══════════════════════════════════════════════════════════════════════

def plot_confusion_matrix(y_test, y_pred, label_map, model_name="Model"):
    """
    Affiche la matrice de confusion du modèle.
    
    La matrice de confusion est un tableau N×N (N = nombre de classes) où :
    - La ligne i représente la vraie classe i
    - La colonne j représente la classe prédite j
    - La case (i,j) indique combien d'images de la classe i ont été classées comme j
    
    → La diagonale = bonnes prédictions, hors diagonale = erreurs
    """
    cm = confusion_matrix(y_test, y_pred)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=list(label_map.values()),
                yticklabels=list(label_map.values()))
    plt.title(f'Matrice de Confusion — {model_name}', fontsize=14, fontweight='bold')
    plt.ylabel('Vraie classe (ground truth)')
    plt.xlabel('Classe prédite')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()


def evaluate_model(model, X_test, y_test, label_map, model_name="Model"):
    """
    Évalue un modèle sur le jeu de test et affiche les métriques.
    
    Métriques affichées :
    - Accuracy    : proportion globale de bonnes prédictions
    - Précision   : parmi les prédictions 'classe X', combien sont vraiment 'classe X' ?
    - Rappel      : parmi les vraies 'classe X', combien ont été correctement détectées ?
    - F1-score    : moyenne harmonique précision/rappel (bonne métrique si classes déséquilibrées)
    - Matrice de confusion : visualisation des erreurs par classe
    
    Retourne:
        accuracy (float), y_pred (numpy.array)
    """
    print(f"\n{'='*50}")
    print(f"  📊 Évaluation — {model_name}")
    print(f"{'='*50}")
    
    # Prédiction sur l'ensemble de test
    y_pred = model.predict(X_test)
    
    # Accuracy globale
    accuracy = accuracy_score(y_test, y_pred)
    print(f"\n🎯 Accuracy : {accuracy:.4f}  ({accuracy*100:.2f}%)")
    
    # Rapport détaillé par classe
    print("\n📋 Rapport de classification (par classe) :")
    print(classification_report(y_test, y_pred,
                                target_names=list(label_map.values())))
    
    # Matrice de confusion
    plot_confusion_matrix(y_test, y_pred, label_map, model_name)
    
    return accuracy, y_pred
```

```python
# ══════════════════════════════════════════════════════════════════════
# PRÉTRAITEMENT : NORMALISATION + PCA
# ══════════════════════════════════════════════════════════════════════

def preprocess_data(X_train, X_test, use_pca=True, n_components=50):
    """
    Applique deux traitements sur les features :
    
    1. NORMALISATION (StandardScaler)
       Ramène chaque feature à une moyenne de 0 et un écart-type de 1.
       → Essentiel pour SVM, qui est très sensible aux échelles des features.
       
    2. PCA — Analyse en Composantes Principales (optionnelle)
       Réduit le nombre de features en gardant celles qui expliquent le plus
       de variabilité dans les données.
       → Accélère l'entraînement et peut améliorer les performances.
    
    ⚠️  IMPORTANT : le scaler et la PCA sont AJUSTÉS (fit) uniquement sur X_train,
        puis APPLIQUÉS (transform) sur X_train et X_test.
        Cela évite la "fuite de données" (data leakage) vers le jeu de test.
    
    Retourne:
        X_train_scaled, X_test_scaled, scaler, pca
    """
    print("⚙️  Prétraitement des données...")
    
    # ── Normalisation ──────────────────────────────────────────────────────────
    # fit_transform sur train : apprend la moyenne/std ET transforme
    # transform sur test      : applique la même transformation (sans réapprendre)
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled  = scaler.transform(X_test)
    print("   ✅ Normalisation appliquée (StandardScaler)")
    
    pca = None
    
    # ── Réduction de dimensionnalité (PCA) ─────────────────────────────────────
    if use_pca:
        print(f"   🔻 Application de PCA → réduction à {n_components} composantes...")
        pca = PCA(n_components=n_components)
        X_train_scaled = pca.fit_transform(X_train_scaled)
        X_test_scaled  = pca.transform(X_test_scaled)
        variance_explained = pca.explained_variance_ratio_.sum()
        print(f"   ✅ PCA appliquée — variance expliquée : {variance_explained:.2%}")
    
    return X_train_scaled, X_test_scaled, scaler, pca
```

```python
# ══════════════════════════════════════════════════════════════════════
# ÉQUILIBRAGE DES CLASSES — SMOTE
# ══════════════════════════════════════════════════════════════════════

def balance_classes_smote(X, y, strategy='auto'):
    """
    Équilibre les classes déséquilibrées en générant des exemples synthétiques.
    
    SMOTE (Synthetic Minority Over-sampling Technique) :
    Pour chaque exemple de la classe minoritaire, SMOTE génère de nouveaux
    exemples en interpolant entre cet exemple et ses voisins les plus proches.
    
    Exemple :
        Avant SMOTE : Classe A = 1000 images, Classe B = 200 images
        Après SMOTE : Classe A = 1000 images, Classe B = ~1000 images
    
    → Utile quand certaines maladies sont beaucoup moins représentées que d'autres.
    
    Retourne:
        X_balanced, y_balanced : dataset équilibré
    """
    from imblearn.over_sampling import SMOTE
    
    print("\n🔄 Application de SMOTE pour équilibrer les classes...")
    print(f"   Distribution avant : {dict(sorted(Counter(y).items()))}")
    
    smote = SMOTE(sampling_strategy=strategy, random_state=42)
    X_balanced, y_balanced = smote.fit_resample(X, y)
    
    print(f"   Distribution après : {dict(sorted(Counter(y_balanced).items()))}")
    print(f"   Taille avant : {len(X)} échantillons")
    print(f"   Taille après : {len(X_balanced)} échantillons")
    
    return X_balanced, y_balanced
```

```python
# ══════════════════════════════════════════════════════════════════════
# ENTRAÎNEMENT — SVM (Support Vector Machine)
# ══════════════════════════════════════════════════════════════════════

def train_svm(X_train, y_train, use_grid_search=False):
    """
    Entraîne un modèle SVM (Support Vector Machine).
    
    Le SVM cherche à tracer le "meilleur hyperplan" qui sépare les classes
    avec la marge maximale. Avec le noyau RBF (Radial Basis Function),
    il peut gérer des frontières de décision non linéaires.
    
    Hyperparamètres importants :
    - C      : paramètre de régularisation. C élevé = modèle complexe (risque surapprentissage)
    - gamma  : influence de chaque exemple d'entraînement. 'scale' = valeur automatique recommandée
    - kernel : type de noyau. 'rbf' est le plus courant pour des données non linéaires
    
    use_grid_search=True : teste plusieurs combinaisons de paramètres pour trouver les meilleurs
    use_grid_search=False : utilise directement les paramètres par défaut (plus rapide)
    
    Retourne:
        model : modèle SVM entraîné (objet sklearn)
    """
    print("\n=== Entraînement SVM ===")
    
    if use_grid_search:
        # GridSearchCV teste toutes les combinaisons de paramètres par validation croisée (cv=3)
        print("🔍 Recherche des meilleurs hyperparamètres (GridSearchCV)...")
        param_grid = {
            'C':      [0.1, 1, 10, 100],
            'gamma':  ['scale', 'auto', 0.001, 0.01],
            'kernel': ['rbf', 'poly']
        }
        svm = SVC(random_state=42, probability=True)
        grid_search = GridSearchCV(svm, param_grid, cv=3, scoring='accuracy',
                                   n_jobs=-1, verbose=1)
        grid_search.fit(X_train, y_train)
        print(f"✅ Meilleurs paramètres : {grid_search.best_params_}")
        print(f"   Meilleur score (CV)  : {grid_search.best_score_:.4f}")
        model = grid_search.best_estimator_
    else:
        # Paramètres par défaut — bons résultats sans recherche exhaustive
        print("⚡ Entraînement avec paramètres prédéfinis (kernel=rbf, C=10)...")
        model = SVC(kernel='rbf', C=10, gamma='scale',
                    random_state=42, probability=True)
        # probability=True permet d'obtenir les probabilités de confiance
        model.fit(X_train, y_train)
        train_acc = model.score(X_train, y_train)
        print(f"   Accuracy sur entraînement : {train_acc:.4f}")
    
    print("✅ Modèle SVM entraîné.")
    return model


# ══════════════════════════════════════════════════════════════════════
# ENTRAÎNEMENT — RANDOM FOREST (Forêt Aléatoire)
# ══════════════════════════════════════════════════════════════════════

def train_random_forest(X_train, y_train, use_grid_search=False):
    """
    Entraîne un modèle Random Forest (Forêt Aléatoire).
    
    Le Random Forest combine plusieurs arbres de décision entraînés sur
    des sous-ensembles aléatoires des données et des features.
    La prédiction finale est la classe la plus votée parmi tous les arbres.
    
    Avantages vs SVM :
    - Plus robuste au surapprentissage
    - Fonctionne bien sans normalisation
    - Peut mesurer l'importance de chaque feature
    
    Hyperparamètres :
    - n_estimators   : nombre d'arbres (plus = mieux, mais plus lent)
    - max_depth      : profondeur maximale de chaque arbre (None = illimité)
    - min_samples_split : nb minimum d'exemples pour diviser un nœud
    
    Retourne:
        model : modèle Random Forest entraîné
    """
    print("\n=== Entraînement Random Forest ===")
    
    if use_grid_search:
        print("🔍 Recherche des meilleurs hyperparamètres (GridSearchCV)...")
        param_grid = {
            'n_estimators':    [100, 200],
            'max_depth':       [10, 20, None],
            'min_samples_split': [2, 5]
        }
        rf = RandomForestClassifier(random_state=42)
        grid_search = GridSearchCV(rf, param_grid, cv=3, scoring='accuracy',
                                   n_jobs=-1, verbose=1)
        grid_search.fit(X_train, y_train)
        print(f"✅ Meilleurs paramètres : {grid_search.best_params_}")
        print(f"   Meilleur score (CV)  : {grid_search.best_score_:.4f}")
        model = grid_search.best_estimator_
    else:
        print("⚡ Entraînement avec n_estimators=200...")
        model = RandomForestClassifier(n_estimators=200, random_state=42)
        model.fit(X_train, y_train)
        train_acc = model.score(X_train, y_train)
        print(f"   Accuracy sur entraînement : {train_acc:.4f}")
    
    print("✅ Modèle Random Forest entraîné.")
    return model
```

```python
# ══════════════════════════════════════════════════════════════════════
# PRÉDICTION SUR UNE IMAGE INDIVIDUELLE
# ══════════════════════════════════════════════════════════════════════

def predict_single_image(image_path, model, scaler, pca, label_map, show_image=True):
    """
    Prédit la maladie d'une plante à partir d'une seule image.
    
    Étapes du pipeline de prédiction :
        1. Extraction des features (comme lors de l'entraînement)
        2. Normalisation avec le scaler ajusté sur le train (si fourni)
        3. Réduction PCA (si fournie)
        4. Prédiction du label par le modèle
        5. Récupération du nom de la maladie via label_map
        6. Affichage de l'image avec la prédiction
    
    ⚠️  Utiliser le MÊME scaler et la MÊME pca que ceux utilisés à l'entraînement.
        Si vous n'aviez pas appliqué de scaler/pca, passez None.
    
    Paramètres:
        image_path : chemin vers l'image à analyser
        model      : modèle entraîné (SVM ou Random Forest)
        scaler     : StandardScaler ajusté (ou None)
        pca        : objet PCA ajusté (ou None)
        label_map  : dictionnaire {numéro → nom de maladie}
        show_image : True pour afficher l'image avec la prédiction
    
    Retourne:
        prediction (int), disease_name (str), proba (float ou None)
    """
    # 1. Extraction des features
    features = extract_features(image_path)
    if features is None:
        raise ValueError(f"❌ Impossible de lire l'image : {image_path}")
    
    features = features.reshape(1, -1)   # Le modèle attend une matrice 2D : (1, nb_features)
    
    # 2. Normalisation (appliquer le MÊME scaler que lors de l'entraînement)
    if scaler is not None:
        features = scaler.transform(features)
    
    # 3. Réduction PCA (appliquer la MÊME pca que lors de l'entraînement)
    if pca is not None:
        features = pca.transform(features)
    
    # 4. Prédiction
    prediction = int(model.predict(features)[0])
    
    # 5. Probabilité de confiance (disponible si probability=True dans SVC, ou avec RF)
    proba = None
    if hasattr(model, "predict_proba"):
        probas = model.predict_proba(features)[0]
        proba  = float(probas[prediction])
    
    # 6. Nom de la maladie (les clés du JSON peuvent être str ou int)
    disease_name = (label_map.get(prediction)
                    or label_map.get(str(prediction))
                    or "Classe inconnue")
    
    # 7. Affichage de l'image avec annotation
    if show_image:
        img_rgb = cv2.imread(image_path)
        img_rgb = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2RGB)   # OpenCV lit en BGR, on convertit en RGB
        
        plt.figure(figsize=(6, 5))
        plt.imshow(img_rgb)
        plt.axis('off')
        
        title = f"Prédiction : {disease_name}\nClasse : {prediction}"
        if proba is not None:
            title += f"  |  Confiance : {proba:.1%}"
        
        plt.title(title, fontsize=13, fontweight='bold', color='darkgreen')
        plt.tight_layout()
        plt.show()
    
    # Affichage console
    print(f"✅ Maladie détectée : {disease_name}")
    print(f"   Classe           : {prediction}")
    if proba is not None:
        print(f"   Confiance        : {proba:.1%}")
    
    return prediction, disease_name, proba
```
